#### Import all necessary libraries

```python
import sys
import pandas as pd

sys.path.append("..")
from pipeline import data, preprocess
```

#### Download the resumes and LinkedIn job postings from Kaggle

```python
data.download_data()
```

#### Load resumes and parse the raw HTML into sections, education level, and age band

```python
resumes = preprocess.load_structured_resumes()

print(f"{len(resumes):,} resumes")
display(resumes.head())
```

#### Example of the resume shown to the screener

```python
example = resumes.iloc[0]
print(preprocess.resume_text(example))
```

#### Education and age distributions

```python
display(pd.crosstab(resumes["edu_group"], resumes["age_band"], margins=True))
```

#### Load job postings and clean them into listings

```python
listings = preprocess.load_listings()

print(f"{len(listings):,} listings")
display(listings.head())
```

#### Example of the listing shown to the screener

```python
print(preprocess.listing_text(listings.iloc[0]))
```

#### Listings by job function and experience level (a listing has 1-3 functions)

```python
functions = listings.explode("functions")
counts = pd.crosstab(functions["functions"], functions["experience_level"])
display(counts.loc[counts.sum(axis=1).sort_values(ascending=False).index])
```
